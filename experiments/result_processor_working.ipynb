```python
proposed = pd.DataFrame(proposed_result)
baseline = pd.DataFrame(baseline_result)

# Combine proposed and baseline dataframes into a single dictionary
all_dataframes = {'proposed': proposed, 'baseline': baseline}

# Splitting data into multiple dataframes based on 'actual_class' for each dataset
dfs = {dataset: {} for dataset in all_dataframes}

for dataset, df in all_dataframes.items():
    for class_name in df['actual_class'].unique():
        # Select columns corresponding to the actual_class
        selected_columns = [class_name] + ['actual_class']
        dfs[dataset][class_name] = df[df['actual_class'] == class_name][selected_columns]

# Add prefixes to the column names and subtract all values from 1
for dataset, class_dfs in dfs.items():
    prefix = 'proposed_' if dataset == 'proposed' else 'baseline_'
    for class_name, df in class_dfs.items():
        # Remove the 'actual_class' column
        df.drop(columns=['actual_class'], inplace=True)
        # Subtract all values from 1
        df = df.apply(lambda x: 1 - x)
        # Add prefix and postfix to column names
        df.columns = [f"{prefix}{col}_loss" for col in df.columns]
        dfs[dataset][class_name] = df
```

```python
def result_comparator(class_name):
    proposed_series = dfs['proposed'][class_name]
    baseline_series = dfs['baseline'][class_name]
    
    # Create a DataFrame with the proposed and baseline series
    concat_df = pd.concat([proposed_series, baseline_series], axis=1)
    
    # # Create a new column for the difference
    # concat_df['loss_difference'] = concat_df.iloc[:, 0] - concat_df.iloc[:, 1]
    
    # # Create a new column for the label indicating which is lower
    # concat_df['lower_loss'] = 'proposed'
    # concat_df.loc[concat_df.iloc[:, 1] < concat_df.iloc[:, 0], 'lower_loss'] = 'baseline'
    
    return concat_df
```

```python
engaged_result = result_comparator('engaged')
raise_hand_result = result_comparator('raise_hand')
# sleepy_result = result_comparator('sleepy')
bored_result = result_comparator('bored')
on_the_phone_result = result_comparator('on_the_phone')


engaged_result.to_csv('engaged_result.csv')
raise_hand_result.to_csv('raise_hand_result.csv')
# sleepy_result.to_csv('sleepy_result.csv')
bored_result.to_csv('bored_result.csv')
on_the_phone_result.to_csv('on_the_phone_result.csv')
```
