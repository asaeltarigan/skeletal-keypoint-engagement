```python
import pandas as pd
import csv
import numpy as np
```

```python
df = pd.read_csv('result_compiled.csv')
```

```python
df = df.loc[:, ~df.columns.str.contains('^Unnamed')]
df
```

```python
df['lower_column'] = np.where(df['proposed'] < df['baseline'], 'proposed', 'baseline')

# Display the DataFrame with the new column
lower_df = df[['proposed', 'baseline', 'lower_column']]

count_lower_column = lower_df['lower_column'].value_counts()

# Display the DataFrame with the new column and count
print("\nAmount of lower cross-entropy loss of proposed method and baseline method:")
print(count_lower_column)
```

```python
import pandas as pd
import numpy as np
from scipy.stats import ttest_rel

# Number of samples (n), mean, standard deviation, and variance for 'proposed'
n_proposed = len(df['proposed'])
mean_proposed = df['proposed'].mean()
std_dev_proposed = df['proposed'].std()
variance_proposed = df['proposed'].var()

# Number of samples (n), mean, standard deviation, and variance for 'baseline'
n_baseline = len(df['baseline'])
mean_baseline = df['baseline'].mean()
std_dev_baseline = df['baseline'].std()
variance_baseline = df['baseline'].var()

# Perform paired t-test
t_statistic, p_value = ttest_rel(df['proposed'], df['baseline'])

# Create a new column 'lower_column' based on the comparison
df['lower_loss'] = np.where(df['proposed'] < df['baseline'], 'proposed', 'baseline')

# Count the occurrences of "proposed" and "baseline" in the 'lower_column'
count_lower_column = df['lower_loss'].value_counts()

# Display the DataFrame with the new column and count
print(df[['proposed', 'baseline', 'lower_column']])

# Display number of samples
print("\nOverall statistics:")
sample_number = n_proposed - n_baseline
if sample_number !=0:
    print(f"There are difference in sample number: {sample_number}")
else:
    print(f"Number of samples (n): {n_proposed}")

# Display statistics for 'proposed'
print("\nStatistics for 'proposed':")
# print(f"Number of samples (n): {n_proposed}")
print(f"Mean: {mean_proposed}")
print(f"Standard Deviation: {std_dev_proposed}")
print(f"Variance: {variance_proposed}")

# Display statistics for 'baseline'
print("\nStatistics for 'baseline':")
# print(f"Number of samples (n): {n_baseline}")
print(f"Mean: {mean_baseline}")
print(f"Standard Deviation: {std_dev_baseline}")
print(f"Variance: {variance_baseline}")

# Display paired t-test results
print("\nPaired t-test results:")
print(f"T-statistic: {t_statistic}")
print(f"P-value: {p_value}")

alpha = 0.05  # significance level
if p_value < alpha:
    print("\nConclusion: Reject the null hypothesis. There is a significant difference between 'proposed' and 'baseline'.")
else:
    print("\nConclusion: Fail to reject the null hypothesis. There is no significant difference between 'proposed' and 'baseline'.")

# Display the number of lower loss in columns
print("\nOccurence of lower loss in 'proposed' and 'baseline':")
print(count_lower_column)

df[['proposed', 'baseline', 'lower_column']].to_csv("count_lower_column.csv")
```
